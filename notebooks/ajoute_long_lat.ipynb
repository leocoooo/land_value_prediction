```python
import pandas as pd
import requests
import time
from tqdm import tqdm

def geocoder_adresse(numero, nom_voie, code_postal, commune):
    """
    Géocode une adresse via l'API Géoplateforme
    """
    # Construire l'adresse complète
    adresse_parts = []
    if pd.notna(numero) and str(numero).strip():
        adresse_parts.append(str(numero))
    if pd.notna(nom_voie) and str(nom_voie).strip():
        adresse_parts.append(str(nom_voie))
    if pd.notna(code_postal) and str(code_postal).strip():
        adresse_parts.append(str(code_postal))
    if pd.notna(commune) and str(commune).strip():
        adresse_parts.append(str(commune))

    adresse = " ".join(adresse_parts)

    if not adresse.strip():
        return None, None

    # Appeler l'API
    url = "https://data.geopf.fr/geocodage/search"
    params = {
        "q": adresse,
        "limit": 1
    }

    try:
        response = requests.get(url, params=params, timeout=10)
        response.raise_for_status()
        data = response.json()

        if data.get("features") and len(data["features"]) > 0:
            coords = data["features"][0]["geometry"]["coordinates"]
            return coords[0], coords[1]  # longitude, latitude
        else:
            return None, None

    except Exception as e:
        print(f"Erreur pour '{adresse}': {e}")
        return None, None


def geocoder_dataframe(df, batch_size=100, delay=0.05):
    """
    Géocode toutes les lignes d'un DataFrame par batch

    Args:
        df: DataFrame avec les colonnes adresse_numero, adresse_nom_voie, code_postal, nom_commune
        batch_size: Nombre de lignes à traiter avant de sauvegarder
        delay: Délai entre chaque requête en secondes (pour respecter la limite de 50 req/s)
    """
    total_rows = len(df)
    print(f"Géocodage de {total_rows} adresses...")
    print(f"Traitement par batch de {batch_size} lignes")

    # Initialiser les colonnes si elles n'existent pas
    if 'longitude' not in df.columns:
        df['longitude'] = None
    if 'latitude' not in df.columns:
        df['latitude'] = None

    # Créer un fichier de sauvegarde
    backup_file = "geocodage_backup.csv"

    # Compter les lignes déjà géocodées
    lignes_a_geocoder = df[df['longitude'].isna()].index
    print(f"{len(lignes_a_geocoder)} adresses à géocoder")

    # Barre de progression
    with tqdm(total=len(lignes_a_geocoder), desc="Géocodage") as pbar:
        batch_count = 0

        for idx in lignes_a_geocoder:
            row = df.loc[idx]

            # Géocoder l'adresse
            lon, lat = geocoder_adresse(
                row['adresse_numero'],
                row['adresse_nom_voie'],
                row['code_postal'],
                row['nom_commune']
            )

            # Mettre à jour le DataFrame
            df.at[idx, 'longitude'] = lon
            df.at[idx, 'latitude'] = lat

            batch_count += 1
            pbar.update(1)

            # Sauvegarder périodiquement
            if batch_count % batch_size == 0:
                df.to_csv(backup_file, index=False)
                print(f"\n✓ Backup sauvegardé ({batch_count} lignes traitées)")

            # Respecter la limite de 50 requêtes par seconde
            time.sleep(delay)

    # Sauvegarde finale
    df.to_csv(backup_file, index=False)
    print(f"\n✓ Géocodage terminé!")

    # Statistiques
    geocoded = df['longitude'].notna().sum()
    print(f"\nRésultats:")
    print(f"  - Adresses géocodées: {geocoded} / {total_rows} ({geocoded/total_rows*100:.1f}%)")
    print(f"  - Adresses non trouvées: {total_rows - geocoded}")

    return df
```

```python
df_geocoded = geocoder_dataframe(
    df,
    batch_size=1000,  # Sauvegarde tous les 100 géocodages
    delay=0.1       # 0.05s entre chaque requête = ~20 req/s (bien en dessous de la limite)
)
```
